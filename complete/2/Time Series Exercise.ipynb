# Group Exercise

In this exercise, you are a team of data analysts at a financial firm. Your task is to analyse the 2022 stock market data of five tech companies. Based on each of your individual analyses, you will draw insights and present your findings as a team. You are given the raw trading data of the following companies in .CSV format:

- AMZN
- META
- MSFT
- NVDA
- TSLA

First, **as a team** develop some code to clean the data in each of the five CSV files. You should handle sorting, missing values, duplicates and correct column dtypes. Apply these rules:

- Forward fill missing Open/Close/Adjusted Close
- Interpolate missing High/Low
- Drop rows with missing Volume

Then, each team member should select a task from the list below. If you have less than five team members, there is no need to complete all tasks - just pick one each. 

You could develop a notebook to solve the task, then run that code on the stocks one at a time. Or for an added challenge, try implementing functions and loops so you can analyse all the stocks at once in a single notebook!

- Calculate the simple daily returns and the cumulative returns for each company. Take note of the average daily return. Also take note of the cumulative return on the last trading day of June, and on the last day of the period.

- Compute the total trading volume for each company for the year. Rank the companies based on this total volume, and report the ranking.

- Calculate the total trading volume of each stock. Then identify the date with the highest trading volume in each quarter. Compare these dates across companies to see if there are any common high-volume trading days.

- Compute the average daily trading range (high price - low price) for each company. Highlight the company with the widest average trading range and the one with the narrowest. Calculate the standard deviation of the daily trading range of each company.

- Determine the days with the largest price jump (the biggest positive difference between the close and opening price) for each company, stating the day and the amount. Do the same for the largest price drop. Discuss the possible reasons for this significant price change based on any notable events or news during that period.

Once everyone is done, come together and discuss your findings. As a group, which one stock would you recommend that your firm invest in, and why?
