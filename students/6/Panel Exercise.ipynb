```python
positive_return_stocks = ["AMD", "BAC", "GE", "MSFT", "NIO"]

def weigh_portfolio(df):
    
    df = df[df['tic'].isin(positive_return_stocks)] # isin will return a boolean df we can use to index the original df
    df_pivot = df.pivot(index="datadate", columns="tic", values="prccd")
    returns = df_pivot.pct_change().dropna()

    def negative_sharpe(weights):
        portfolio_returns = np.dot(returns, weights)
        annualised_returns = portfolio_returns.mean() * 252
        annualised_volatility = portfolio_returns.std() * np.sqrt(252)

        sharpe_ratio = (annualised_returns - 0.05) / annualised_volatility

        return -sharpe_ratio
    
    my_cons = {"type": "eq", "fun": lambda x: sum(x) - 1}
    my_bounds = tuple((0, 1) for x in range(5))
    init_weights = np.ones(5) / 5

    optimal = minimize(negative_sharpe, init_weights, bounds=my_bounds, constraints=my_cons)
    print(optimal)


weigh_portfolio(df)
```
