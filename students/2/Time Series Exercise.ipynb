# Group Exercise

In this exercise, you are a team of data analysts at a financial firm. Your task is
to analyse the 2025 stock market data of five food service companies. Based on each
of your individual analyses, you will draw insights and present your findings as a
team. You are given the raw trading data of the following companies in .CSV format:

- MCD (McDonald's)
- SBUX (Starbucks)
- CMG (Chipotle)
- DPZ (Domino's Pizza)
- WEN (Wendy's)

First, **as a team** agree on the code to clean the data, so that all five files
are treated the same way. You should handle sorting, missing values, duplicates and
correct column dtypes. Apply these rules:

- Forward fill missing Open/Close
- Interpolate missing High/Low
- Fill missing Volume with the median volume, remembering it counts shares

Then, each team member should take one company from the list above. If you have
less than five team members, there is no need to cover all five - just take one
each. You should each end up with 249 trading days and no missing values.

Now work out the following for your own company, and write the figures down
somewhere your team can see them:

- The number of trading days you ended up with
- The close on the first and the last trading day, and the change between them
- The highest and the lowest close, and the dates they happened
- The average daily trading range (high price - low price), both in dollars and as
  a share of that day's closing price
- The busiest trading day by volume, and the date
- How many days your company closed above its own average close

You could work through these one at a time. Or for an added challenge, try
implementing a function so you can produce the whole set in one go!

Once everyone is done, come together and discuss your findings. Collect your
figures into a single table, with a row for each company, and compare your average
daily trading ranges - first in dollars, then as a share of the price. Do you get
the same ranking both ways, and which would you show a client? As a group, which
one stock would you recommend that your firm invest in for the coming year, and why?
