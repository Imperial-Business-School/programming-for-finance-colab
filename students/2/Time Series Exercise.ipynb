# Group Exercise

In this exercise, you are a team of data analysts at a financial firm. Your task is
to analyse the 2025 trading data of five food service companies. You have about 40
minutes.

Take one company each:

- MCD (McDonald's)
- SBUX (Starbucks)
- CMG (Chipotle)
- DPZ (Domino's Pizza)
- WEN (Wendy's)

With fewer than five people, take one each and leave the rest. You will each do the
same short piece of work, then bring your numbers together at the end.

## Cleaning the data

Your file arrives unprocessed. Sort it, handle the duplicate records, and correct
the column dtypes. Then deal with the missing values. **Everyone should use the
same rules here**, so that the results can be compared afterwards:

- Forward fill missing `Open` and `Close`
- Interpolate missing `High` and `Low`
- Fill missing `Volume` with the median volume, and keep it a whole number

You should end up with 249 trading days and nothing missing. Check both. One of
these five files has a gap the rules above will not close, so look rather than
assume. If that one is yours, decide what to do about it and be ready to say why.

## Describing your company's year

Work out the following, and write them down where your team can see them.

- The number of trading days you ended up with
- The close on the first and the last trading day, and the change between them
- The highest and the lowest close, and the dates they happened
- The average daily trading range - `High` minus `Low` - in dollars, and again as a
  share of that day's close
- The busiest trading day by volume, and the date
- How many days the company closed above its own average close

## Bringing it together

Collect everyone's figures into one table, a row for each company. Which company
had the strongest year? Which had the weakest?

Now look again at the average daily ranges. Rank them in dollars, then rank them as
a share of the price. Do you get the same answer? If not, which ranking would you
put in front of a client?

Did any two companies have their busiest day on the same date? What might explain
that?

Finally, agree a recommendation. If the firm could hold just one of these five for
the coming year, which would you choose? What in your numbers supports it, and what
does your analysis *not* tell you?

### If you finish early

- Break the figures down by quarter and see whether the picture changes
- Compare the first half of the year with the second
- See whether any weekday is consistently busier than the others
